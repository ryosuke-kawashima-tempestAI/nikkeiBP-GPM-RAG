```python
import pandas as pd
import os
import datetime
import json
```

```python
def files_in_dir(directory_path):
    files_and_dirs = os.listdir(directory_path)
    print(files_and_dirs)

def get_current_datetime_components():
    """Returns the current year, month, day, hour, and minute as formatted strings."""
    now = datetime.datetime.now()
    return f"{now.strftime("%Y")}-{now.strftime("%m")}-{now.strftime("%d")}-{now.strftime("%H")}-{now.strftime("%M")}"
```

```python
TARGET_PATH = "./target/ma_welding_llds.xlsx"
target_json = json.load(open("./outputs/result-2026-01-04-13-20.json", "r"))
df_lld_data = pd.read_excel(TARGET_PATH)
```

```python
def show_keys(json_data: dict, depth: int = 0) -> dict:
    keys = {f"Depth {depth}": json_data.keys()}
    for key, value in json_data.items():
        if isinstance(value, dict):
            keys.update(show_keys(value, depth + 1))
    return keys
```

```python
print(target_json["lld_grouping"]["ClassID"])
```

```python
df_lld_data["ClassID"] = target_json["lld_grouping"]["ClassID"]
df_lld_data["ClassificationReason"] = target_json["lld_grouping"]["ClassificationReason"]
df_gpm_data = pd.DataFrame(target_json["gpm_reconstruction"])
# Use map or merge(on="ClassID", how="left")
merge_data_partof = pd.DataFrame({"ClassID": [i for i in range(1, len(target_json["lld_grouping"]["PartOf"])+1)], "PartOf": target_json["lld_grouping"]["PartOf"]})
merge_data_partof_reason = pd.DataFrame({"ClassID": [i for i in range(1, len(target_json["lld_grouping"]["PartOfReason"])+1)], "PartOfReason": target_json["lld_grouping"]["PartOfReason"]})
df_gpm_data = df_gpm_data.merge(merge_data_partof, on="ClassID", how="left")
df_gpm_data = df_gpm_data.merge(merge_data_partof_reason, on="ClassID", how="left")

lld_output_path = f"./outputs/ma-welding-lldgrouping-{get_current_datetime_components()}.xlsx"
gpm_output_path = f"./outputs/ma-welding-gpmreconstruction-{get_current_datetime_components()}.xlsx"
df_lld_data.to_excel(lld_output_path, index=False)
df_gpm_data.to_excel(gpm_output_path, index=False)
print("LLD Data was sucessfuly saved!")
```
